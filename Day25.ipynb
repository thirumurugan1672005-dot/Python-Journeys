# Assignment makes references 

* Remember Assignment makes references
* Slicing , concatenation , repetition ,copy() method top level copies
* # Repetition makes nested copies like concatenation , slicing and copy() method

```python
a = 12 
b = a
print(b) # fine
```

```python
# mutable 
L =[1,2,3]
M = L
L[0] = 24
print(M)
```

```python
# slicing , concatenation (nested shared references)
K = [1,[2,3],34]
M = K[:]
E = K + [1,2,3]
print(K,M,E)
K[0] = 24
print(M,E)
K[1].append(100)
print(M,E)
```

```python
# Repetition also shares nested copies
Y = [1,2,4]
E = Y * 2
Y[0] = 100
print(E)

X = [1,[2,3,4],45]
Z = X * 2
X[1][0] = 23
print(Z)
```

```python
# here the reflects changes
X = [1,2,3]
Y = [X] * 2
print(Y)
Y[0][1] = 100
print(X)
# avoid it
Y = [list(X)] * 2
Y[0][1] = 23
print(X)
```
