## Escape Sequences
* Escape Sequences are the sequences which is used to represent characters which we cannot type or represent
* The word / after one or more characters gets encoded into single character in resulting string

```python
# /n : newline
exp  = "Hello\nWorld"
print(exp)
exp
```

```python
# /t : tabspace
exp = "Hello\tWorld"
print(exp)
exp
```

```python
# len : length of string 
exp = "Hello\tWorld"
len(exp)
```

```python
# How to get code point for the string
ord("\n")
```

```python
# How to code point of tabspace
ord("\t")
```

<p>Generally length is not same as bytes it refers to number of code points <br/>
Single Code point can be hold more than one byte <br/> Unicode system Code point says
that which character should be used<br/>Ascii characters has exactly one byte others may have more than one byte </p>

```python
exp = "p".encode('utf-16')
print(exp)
print(len(exp))
exp = "p".encode('utf-8')
print(exp)
print(len(exp))
```

## Raw Strings
* In Window file paths and Regex Strings there is combination of backslash
* In order to switch off the escape mechanism we use Raw Strings
* They have r'Python'
* Nevertheless , They cannot end with odd pairs of backslashes since /' is not valid

```python
# error
window_path = "C:\new\Path"
print(window_path)
```

```python

window_path = r"C:\new\Path"
print(window_path)
```
