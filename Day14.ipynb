# Lists
* Arbiratory collection contains anything like number , strings and even other lists
* Lists are sequences can be accessed by offsets and concatenation , repetition and all the sequence operations work
* Lists are heterogenous and can contain any types
* Lists supports duplicate elements
* Lists are ordered from offsets zero to length of list - 1
* Lists are mutable as we can change in place either by offsets or by slicing
* Lists can also be nested

* Lists : coded series of objects enclosed by square brackets seperated by commas
* Lists : [] -> empty list
* Implemented as arrays in CPython Interpetor

# Basic List Operations
1. Concatenation (+)
2. Repetition (*)

```python
# empty list 
empty_list = []
print(empty_list)
# list 
arb_list = [1,"name",True,3.14,[1,2,3]]
print(arb_list)
# concatenation
print([1,2,3] + [4,5,6])
# repetition
print([1,2,3]*3)
```

* List cannot concatenate with string directly throws TypeError
* List also be compared left to right like strings

```python
['name']+'name'
```

```python
['name']+list('name')
```

```python
[1,2,3] == [1,2,3]
```

```python
[1,[3,4],5] == [1,[3,4],5]
```

```python
[2,3,4,5] < [1,2,3,4,5]
```

## Indexing and Slicing

```python
a = [1,2,3,4]
print(a[0],a[-1])
print(a[len(a)-1])
```

```python
# index error
a = [1,2]
a[3]
```

```python
a = [1,2,3,4,5,6,7]
print(a[:4])
print(a[2:])
print(a[1:5])
print(a[::-1])
print(a[:])
```

```python
a = [1,2,3,4,5]
print(a[-1:2:-1])
```

```python
mat = [[1,2,3],[4,5,6]]
print(mat[1][2])
```

# InPlace Changes 
* Index Assigning
* Slice Assigning

```python
# in place change : change within object
a = [1,2,3]
a[0] = 100
print(a)
```

```python
# slice change : change slice of object
a = [1,2,23,3]
a[:2] = [12,100,12]
print(a)
```

Slicing Changes occurs because 
* Python first deletes the slice
* and inserts the element at place where slice used to be

```python
a = [1,2,3,4]
a[1:1] = [100]
print(a)
```

```python
a = [1,2,3,4,5]
a[1:] = []
print(a)
```
