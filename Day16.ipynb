# Dictionaries
* Dictionaries are accessed by keys not offsets
* Dictionaries are mutable and grow and shrink
* Dictionaries are mappings in which they map key to values
* Dictionaries maintain insertion order from python 3.6
* Dictionaries are not sequences instead they are mappings like hash table
* Retreival keys is faster

```python
# empty dictionary
empty_dict = {}
print(empty_dict)
# dictionary
user = {"username":"user","password":"PASSKEY"}
print(user)
# keys 
print(user.keys()) # iterable
print(list(user.keys()))
print(user['username'])
```

* in membership works only checks the keys
* len() : return the number of keys in dictionary

```python
user = {"username":"user","password":"PASSKEY"}
print(user)
print("username" in user)
print("user" in user)
print(len(user))
```

## Changing Dictionary in place
* del : del entire dictionary
* del dict[key] : delete particular key
* pop(key) : remove the particular key
* dict[key] = name : if already present update or create new key

```python
user = {}
user['name'] = 'sample'
user['password'] = 'sample@123'
user['name'] = user['name'].title()
print(user)
user['std'] = 123
print(user)
user.pop('std')
print(user)
del user['name']
print(user)
```

```python
fruits = {"apples":250,"bananas":50,"oranges":240}
if 'apples' in fruits:
     del fruits['apples']
```

```python
# Presence of nested Dictionaries
person = {
    "username" : {
        "first_name":"sample",
        "last_name":"user"
    },
    "scores" : [100,200,300]
}
print(person['username']['first_name'])
print(person['scores'][1])
```
