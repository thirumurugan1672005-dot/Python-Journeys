# Bitwise Operators
Bitwise Operators operators on bits level

```python
# Shift Operators (Left Shift)
a = 0b100
a =  a << 2 
bin(a)
```

```python
# Shift Operator (Right Shift)
a = 0b101
a = a >> 1
bin(a)
```

```python
# OR (Bitwise)
a = 23
b = 12
print(a | b)
print(f"a = {bin(a)}")
print(f"b = {bin(b)}")
print(f"a | b = {bin(a|b)}")
```

```python
# AND (Bitwise)
a = 23
b = 12
print(a & b)
print(f"a = {bin(a)}")
print(f"b = {bin(b)}")
print(f"a & b = {bin(a&b)}")
```

```python
# XOR (Bitwise)
a = 23
b = 12
print(a ^ b)
print(f"a = {bin(a)}")
print(f"b = {bin(b)}")
print(f"a ^ b = {bin(a^b)}")
```
