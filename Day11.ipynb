## Case change things
1. captialize
2. upper
3. lower
4. swapcase
5. title

```python
help("string".capitalize)

# example
sentence = "hello world"
print(sentence.capitalize())

help("string".title)

# example
sentence = "today wilson stand by watch your show"
print(sentence.title())

help("string".upper)

# example
print(sentence.upper())

help("string".lower)

# example
sentence = "MARY"
print(sentence.lower())

help("string".swapcase)

# example
sentence = "mAstEr"
print(sentence.swapcase())
```
