# Dictionary key/items/value view objects 
* they returns view objects
* since keys and items view can be hashable they support set operations

````python
D = {'a':1,'b':2}
print(D)
print(D.keys())
print(type(D.keys()))
````

````python
D = {'a':1,'b':2,'c':3}
print(D.values())
print(type(D.values()))
````

````python
D = {'a':1,'b':2,'c':3}
print(D.items())
print(type(D.items()))
````

````python
D | D.values()
````

````python
E = {'d':2,'e':1200,'a':12}
print(D | E)
print(D | E.keys())
print( E | D)
print(E.keys() | D)
print(type(E.keys() | D))
````

````python
E.keys() | D.keys()
````

````python
E.keys() & D.keys()
````

````python
E.items() | D.items()
````

````python
E.items() & D.items()
````

# Sorting Dictionary 
* keys of dictionary can be sorted by sorted() which takes iterable
* or converts into list and use sort method
# Magnitude comparisons 
*  we can use == to check dictionaries equal or not
*  ```> or < ``` not manually work on dictionaries but they work on item view

````python
D = { i-97 : chr(i) for i in range(97,123)}
print(D.keys())
````

````python
E = {i-97:chr(i) for i in range(97,123)}
D == E
````

````python
D != E
````

````python
D > E # unsupported
````

````python
D.items() > E.items()
````

````python
D.values() == E.values()
````

````python
D.values() < E.values()
````

````python
D.keys() < E.keys()
````

````python
# keys and item view support < and >
````

````python
D.items() == E.items()
````

````python
D.keys() == E.keys()
````

````python
# sort dictionary
for i in sorted(D.keys()):
    print(D[i],end = " ")
````

````python
l = list(D.keys())
l.sort()
for i in l:
    print(D[i],end = ' ')
````
