## Operator Precedence
* Operators are arragned in higher to lower precedence
* Operators have tightly bound to objects have higher precedence
* Operators have loosely bound to objects have lower precedence
* Operators with high precedence execute first
* Operators with same precedence execute left -> right except Exponentation

````python
2**2**3
````

````python
2**8
````

````python
2*3/3
````

````python
expr = 2 * 4 + 3 # executed as 2*4 and then add 3
print(expr)
````

## Parthensis Helps to Override Precedence rules
It also ables to readability and understand better

````python
expr = (2 * 4) + 3 # Now We sure 2*4 executes first and then add 3
print(expr)
expr = 2 * (4 + 3) 
print(expr)
````
