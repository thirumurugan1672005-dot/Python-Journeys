```python
# using conversion tools 
file = open("conv.txt",'w')
file.write('Hello\n')
file.write('2,3,4,5\n')
file.write('[1,2,3]${"name":"user","age":32}')
file.close()
```

```python
file = open("conv.txt",'r')
sen1 = file.readline()
print(sen1)
sen2 = file.readline()
l = sen2.rstrip().split(',')
l = [int(i) for i in l]
print(l)
sen3 = file.readline()
L , D = sen3.rstrip().split('$')
L = eval(L)
D = eval(D)
print(L,D)
print(type(L),type(D))
```

```python
# using pickle module 
file = open('pick.pkl','wb')
import pickle
D = {'apple':320,'orange':240}
pickle.dump(D,file)
file.close()
```

```python
D = {'name':'sample','age':32}
import pickle 
F = open("sample.pkl",'wb')
pickle.dump(D,F)
F.close()
```

```python
# json
user = {
    'name' : 'sample',
    'age' : 23
}
file = open("user.json",'w')
import json
json.dump(user,file)
file.close()
```

```python
file = open("user.json",'r')
file.read()
```

```python
file.seek(0)
ans = json.load(file)
print(ans,type(ans))
```

```python
# use csv
import csv

rows = [
    ['name', 'age', 'city'],
    ['Ann', 30, 'Chennai'],
    ['Bob', 25, 'Delhi'],
]

with open('people.csv', 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(['name', 'age', 'city'])   # one row
    w.writerows(rows[1:])                 # many rows
```

```python
with open("people.csv" ,'r') as f:
    w = csv.reader(f)
    for line in w:
        print(line)
```

# File context managers
*  with statements make sure file close no matter what
*  scope is the where file can be accessed
*  Automatically close resources even when code interrupts

```python
with open("file.txt",'r') as f:
    print(f.read())
```
