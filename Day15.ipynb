## List Methods 
* append : append the element at end
* extend : extend the list by accepting iterable
* insert : insert element at particular position or at end if it is out of bounds
* The above methods inplace changes

```python
a = []
a.append(1)
a.append('hack')
print(a)
a.extend([1,2,3])
print(a)
a.insert(0,12)
a.insert(100,12)
print(a)
```

* sort : sort the list inplace
* sorted : return sorted list not inplace

```python
num = [200,2,300,4]
num.sort()
print(num)
```

```python
num = [200,2,300,4]
x = sorted(num)
print(num)
print(x)
```

sorting does not support between types we cannot compare magnitudes

```python
num = [1,1.2,3,4.5]
num.sort()
print(num)
```

```python
num = ["Python","Ava","Long"]
num.sort()
print(num)
```

```python
num = [1,"Python",12.3]
num.sort()
```

```python
# It applies functions to the all the elements avodinf error
num = [1,"Python",12.3]
num.sort(key = str)
print(num)
```

```python
# reverse : used to descending order
num = [1,2,3,4]
num.sort(reverse=True)
print(num)
```

## Other List Methods 
* remove() : remove the element from list if not found ValueError
* index() : find the index of element in list if not found throw ValueError
* pop() : pop the element based on index default:last element
* pop() : used in out of range or and empty list throws Index Error

```python
a = [1,2,3]
a.remove(2)
print(a)
a.pop()
print(a)
```

##  Iteration , Comprehension and Unpacking

```python
num = [1,2,3]
for n in num:
    print(n)
```

```python
num = [1,2,3]
for n in range(len(num)):
    print(num[n])
```

```python
num = [i ** 2 for i in range(5)]
print(num)
```

```python
num = [i for i in range(100) if i % 5 == 0]
print(num)
```

```python
num = [1,2,3]
print(*num)
```

```python
num = [1,2,3,4,5,6]
first , *middle , last = num
print(first , last)
print(middle)
```

* clear : clear the entire list
* copy : copy the list
* del : del by offsets or remove variable itself

```python
num = [1,2,3]
num2 = num.copy()
print(num,num2)
num.clear()
print(num)
```

```python
num = [1,2,3,4,5]
del num[2:]
print(num)
```

```python
num = [1,2,3,4]
del num[2]
print(num)
del num
print(num)
```

```python
num =[1,2,3]
num[:] # identical of copy
```

```python
# Affecting elements 
num = [1,2,3]
num1 = num
num1[0] = 100
print(num)
```

```python
# use copy so it does not affect
num = [1,2,3]
num1 = num[:]
num1[0] = 100
print(num)
```
