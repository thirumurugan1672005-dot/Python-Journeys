* From Python 3.6+ Dicitonary maintaines insertion order
* newly added keys always add at the end
* we cannot add keys in middle in dictionary
* update keys dont change order

```python
fruits = {}
fruits['apple'] = 220
fruits['grapes'] = 180
print(fruits)
fruits['apple'] = 240
print(fruits)
```

```python
# popItem() : removes last inserted key
fruits = {}
fruits['apple'] = 220
fruits['grapes'] = 180
print(fruits)
fruits['apple'] = 240
print(fruits)
print(fruits.popitem())
print(fruits)
```

UPDATE ,COPY , | OPERATORS

```python
fruits = {}
fruits['apple'] = 220
fruits['grapes'] = 180
print(fruits)

fruits_copy = fruits.copy()
fruits_copy.update({'bananas':35})
print(fruits_copy)

new_fruits = fruits | {'bananas':35}
print(new_fruits)
```

# Avoiding Missing keys for dictionaries
1. use in operator to check before using if statement
2. use try-catch to catch errors
3. use get() methods to assign default values

```python
Matrix = {}
Matrix[(0,1,2)] = 78
Matrix[(2,1,4)] = 88
Matrix[(1,2,3)] = 98
print(Matrix[(0,1,2)])
print(Matrix.get((0,1,2),0))
print(Matrix.get((0,3,2),0))
```

# Nesting in Dictionaries

```python
person = {
    'name':{
        'first_name' : 'Ram',
        'last_name' : 'Kumar'
    },
}
person['name']['first_name']
```
