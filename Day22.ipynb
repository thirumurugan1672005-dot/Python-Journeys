## File Iterators 
We can iterate file object in for loops 

* b : processing binary files
* ```+```: open the file for both input and output

````python
file = open('folder/file.txt','w')
file.write('im waiting')
file.close()
````

````python
file = open('folder/file.txt','r')
sen = file.read()
file.close()
sen
````

````python
file = open('folder/file.txt','w')
file.write('Hello')
file.close()
# We can use iterate files using for loops
for line in open('folder/file.txt','r'):
    print(line)

file = open('folder/file.txt' , 'w')
file.write('World')
file.close()
# over write
for line in open('folder/file.txt','r'):
    print(line)

# now append
file = open('folder/file.txt' , 'a')
file.write('World')
file.close()
# over write
for line in open('folder/file.txt','r'):
    print(line)
````

````python
file = open('file.txt','r+')
file.read()
````

````python
file.write('Today')
file.read() # End of Line
````

````python
# binary mode 
bin_file = open('sample.txt','w')
bin_file.write('Hellobin')
bin_file.close()
bin_file = open('sample.txt','rb')
for line in bin_file:
    print(line)
````

* Contents should be strings not objects
* close() method is used to close the resources

````python
file = open('folder/file.txt','w')
file.write(12)
file.close()
````

````python
# above example file may not be closed 
with open('folder/file.txt','r+') as f:
    f.write(str(12))
    f.seek(0) # seek the jump to the position
    print(f.read())
````
