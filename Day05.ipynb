# Arithemetic Operators
* (+) - Addition
* (-) - Subraction
* (*) - Multiplication
* (/) - Division
* (//) - Floor Division
* (%) - Modulus
* (//) - Floor Division

```python
# Addition
a = 12 + 34
print(a)
# Subraction
b = 234 - 13
print(b)
# Multiplication
c = 12 * 13
print(c)
# Division
d = 12 / 144
print(d)
# Modulus
e = 144 % 12
print(e)
# Exponentation
print(12 ** 2)
```

# Mixed Types are used
* When doing Operations Mixed Types between Numbers it converts to complicated types and done that operation
* Complicated Types : Complex > Floating > Integer
* If either of operands are complex it is complex
* else if either of operands are float it is float
* else it is integer
* However it works between only numbers
* You cannot do operations between integer and string (+) results in Type Error

```python
a = complex(12,3) + complex(2,3)
print(a)
a = a + 3.2
print(a)
a = a - 2
print(a)
```

```python
a =  2.3 + 2
print(a)
a = 2 + 3
print(a)
```

# Operator Overloading 
Operators do operations based on types
* (+)  addition for numbers
*  (+) : concatenation for strings
* It is called operator overloading
* Operators overloaded to give operations

# Divison Operator
* / : Division Operator : It results always floating point numbers
* <p> // : Floor Division Operator : Floor Division done floor of result and type based on operand types </p>
* % : Modulus Operator : Give  Remainder
* divmod() function used to get both quotient and remainder

```python
a = 12 / 3
b = 12 / 45
print(a,b)
```

```python
c = 39 // 13
d = 45 // 34.23
print(c,d)
```
