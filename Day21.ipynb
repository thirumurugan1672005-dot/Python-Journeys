# Tuples 
* Tuples are immutable form of lists
* in which they cannot change in-place
* They cannot grow or shrink
* They contain heterogenous things but immutability is in top level
* nested objects may mutate

```python
# empty and single tuple
e = tuple()
print(e)
f = ()
print(type(f))
g = (5) # dont forget commas
print(type(g))
h = (5,)
print(type(h))
```

```python
# tuple assignment is forbidden
t = (1,2,3,4)
t[0] = 20
```

```python
# Access by offsets and supports slicing
t = (1,2,3,4,5)
print(t[0],t[-1])
print(t[::-1],t[2:4])
```

```python
# Concatenation and Repetition
t = 1,2,3,4
e = 5,6,6,7
t + e
```

```python
(1,2,3,4,5) * 2
```

* Parthensis : within function call  , nested in large expressions
* commas : within function call or embeded in other structures
* commas is what it makes tuples 

```python
a = 1 , 2
print(a)
```

```python
# unpacking 
t1 = 'python' , [1,2,3] ,('master','gh'),True
s1 , *t = t1
print(s1,t) # unpacked into list 
t1 = "first","second",1,2,"last"
a1 , a2 , *mid , l = t1
print(a1,a2,l)
```

# Sorting tuple 
* we can convert into list and sort and then make it as tuple
* sorted() to generate new tuples

```python
t = ('a','b','c')
t.index('a')
```

```python
t = ('a','b','c')
t.index('d')
```

```python
t = ('a','b','c')
t.count('a')
```

```python
# sort tuple 
tup = (12,34,123,23,45)
tup = list(tup)
tup.sort()
tup = tuple(tup)
print(tup)
```

```python
# sort tuple 
tup = (12,34,123,23,45)
tup = tuple(sorted(tup)) # sorted returns list
print(tup)
```

```python
# tuple comprehension IS NOT EXISTED WE MUST CONVERTED INTO TUPLE
x = tuple(i for i in range(10))
x
```

Nested objects in tuple can be mutated

```python
t = ([1,2,3],'python',[23])
t[2].append(45)
t[0][1] = 200
print(t)
```

## NamedTuples
* NamedTuples are tuples we can access both name and positions

```python
pat = dict(name="User",email="User@gmail.com",positions=["dealer","councilor"])
print(pat)
offset_pat = tuple(pat.values())
print(offset_pat)
print(pat['name'])
print(offset_pat[0])
```

```python
from collections import namedtuple
p = namedtuple("R",['name','age','jobs'])
p = p(name='User',age=35,jobs=['dev','artist'])
print(p[0],p.name)
```

```python
p_dict = p._asdict()
print(p_dict['name'])
```
