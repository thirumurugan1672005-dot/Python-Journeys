# More Dictionary Methods 
* Dictionary Methods provide a variety of type specific tools
* keys() : return keys in dictionary
* values() : values in dictionary
* items() : (key,value) pairs in dictionary
* They return iterables and list() makes them to list

```python
students = {
    "geetha" : 98,
    "max" : 78,
    "george" : 67
}
print(students)
print(students.keys())
print(students.values())
print(students.items())
print(list(students.keys()))
print(list(students.values()))
print(list(students.items()))
```

* Fetching key if not present results in keyerror
* we can use get method to give default values if not present

```python
fruits = {"papaya":65,"banana":35,"apple":240}
print(fruits)
print(fruits['papaya'])
print(fruits.get('papaya','unknown'))
print(fruits.get('grapes','unknown'))
```

* Update method : makes to update with other dictionary simply overwrite the existing keys and create if they are newones
* pop(key) : pop based on keys
* copy() : copy entire dictionary

```python
fruits = {}
fruits1 = {'apple':240,'grapes':180,'banana':35}
fruits2 = {'grapes':200,'guava':65}
fruits.update(fruits1)
print(fruits)
fruits.update(fruits2)
print(fruits)
fruits.pop('grapes')
print(fruits)
```

```python
fruits = {"papaya":65,"banana":35,"apple":240}
fruits.pop('grapes')
```

```python
fruits = {"papaya":65,"banana":35,"apple":240}
fruits_x = fruits 
print(fruits)
fruits_x['apple'] = 260 # reflects
print(fruits)
```

```python
fruits = {"papaya":65,"banana":35,"apple":240}
fruits_x = fruits.copy()
print(fruits)
fruits_x['apple'] = 260 # r not eflects
print(fruits)
```
