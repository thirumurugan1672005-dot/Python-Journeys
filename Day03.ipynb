# Integer Precision
Integer Precision has unlimited precision but it has peformance penalty when there are more digits in an operation

```python
a = 2 ** pow(10,6)
print(a)
```

```python
2 ** pow(10,4)
```

### Complex numbers
* Complex numbers are numbers with imaginary portions
* They have two parts real and imaginary while real is optional
* imaginary parts are represent as j or J suffix in python
* They are represented as two floating point numbers
* They can also created by complex(real,img)

```python
c = 12 + 35j
print(c)
d = 345J
print(d)
```

```python
e = complex(12)
print(e)
f = complex(45,56)
print(f)
```
