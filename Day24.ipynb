# Reference and Copying
* when you give the value of variable , you are giving reference
* If the Shared References of mutable type , inplace change reflects as the both of denotes same object

```python
a = 2
b = a
print(a,b)
```

```python
a = [1,2,3]
b = a
b[0] = 23
print(a)
```

```python
# copy 
a = [1,2,3]
b = a[:]
a[0] = 23
print(b)
```

```python
# slcing only copies top level copies
L = [[1,3,3],[4,5,6]]
M = L[:]
M[0][0] = 100
print(L)
```

```python
# copy() method 
D = {'name':'Karen','age':32}
E = D
E['name'] = 'Mukesh'
print(D)

F = {'name':"Karen",'age':32}
G = F.copy()
F['name'] = 'Mukesh'
print(G)
```

```python
# copy() method does not save for sharing nested references
D = {'name':{'first_name':'Mukesh','last_name':'Ambani'}, 
     'age': 32
    }
E = D.copy()
E['name']['first_name'] = 'Sukesh'
print(D)
```

```python
# deepcopy
from copy import deepcopy
D = {'name':{'first_name':'Mukesh','last_name':'Ambani'}, 
     'age': 32
    }
E = deepcopy(D)
E['name']['first_name'] = 'Sukesh'
print(D)
```

# Comparison and Equality
* == checks for equivalence of values and go recursively check nested objects left to right
* is checks whether the two objects are same 

```python
# Numbers (check by maginitude)
print(12 == 13)
print(12 > 1)
print(12 < 1)
print(12 >= 1)
print(12 <= 12.32)
```

```python
# Strings (check lexographically)
print('hack' > 'apple')
print('apple' == 'apple')
print('zebra' <= 'zebf')
```

```python
# Sets (Dont care about order)
print({1,2,3} == {3,2,1})
```

```python
# Lists , Tuples (check nested forms and elements left to right)
l = [[1,2],[2,3]]
m = [[1,2],[2,3]]
x = m + [[2,3]]
print(l==m)
print(tuple(l) == tuple(m))
print(x)
print(x <= m)
```

```python
# Dictionary 
D = {'name':'Kevin','age':32}
E = {'name' : 'Kevin','age':32}
print(D == E)
print(D >= E) # not supported
```

```python
# Dictionary 
D = {'name':'Kevin','age':32}
E = {'name' : 'Kevin','age':32}
print(D.keys() >= E.keys())
print(D.items() >= E.items())
print(D.values() >= E.values())# not supported for dict_values
```

```python
# non numeric type 
'1' > 1 # error
```

## Dictionary Comparison
Dictionary does not support magnitude comparisons

```python
# but items() and keys() supports it 
D = {'apple':240,'oranges':180}
E = {'apple':240,'oranges':290}
print(D.items() < E.items())
print(D.keys() < E.keys())
```

```python
sorted(D.items()) < sorted(E.items())
```
