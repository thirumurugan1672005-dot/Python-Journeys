# <b>String formatting </b>
* String formatting is the way to format results and delivered to user
* It is one of the task tells how we represent the results

## <b>Formatting using % operator <b>

```python
exp = "My Age is %d" % 20
print(exp)
```

From Above code 
* on the left of % there is string which has %d denotes integer(format string)
* on the right of % there is single value or a tuple which replaces %d(format string)
* % operator works as formatting operator

```python
exp = "My name is %s, His Purse has %s rupees" % ("Ravi",2000)
print(exp)
```

* Any Value can be converted into string
* Strings are immutable so Formatting creates only new String
* It does not change on left

```python
"%f" % (3.5)
```

## Format Strings
1. %d or %D : integer
2. %x or %X : hexadecimal
3. %o or %O : octal
4. %f or %F : float
5. %E or %e : float with exponent

```python
"%e" % 300.56
```

%[keyname][flags][width][precision]type
* [] are optional
* precision : set the numbers of digits after decimal in floating point numbers

```python
"%.2f" % 35.5678
```

width : number of digits

```python
len("%05d" % 567) # giving minimum width 
print("%5d" % 567)
```

```python
# numeric sign (+) get correct sign
print("%+08.2f" % -32.56)

# zero padding
print("%05.3f" % 3.5)
```

```python
# take width and precision dynamically using *
"%0*.*f" % (6,2,32.44)
```

```python
# using dictionary keys
#{key : value} from format

"My name is %(name)s and age is %(age)s ." % {"name":"Joker","age":25}
```

## <b>Format method </b>
* It creates formatted string from Python 3 version
* It uses named , position , absolute value arguments for formatting
* It subsituted in the {} to create formatted string
* It creates new string not change in place since it is immutable

<b> using  positional arguments order in which they arrange

```python
"My name is {} My age is {}".format("Joker",25) #  replaced with positional arguments
```

<b> using Absolute value postions from 0 to len(arguments) -1

```python
"My name is {1} My age is {0}".format(25,"Joker") # absolute position
```

<b> use named arguments

```python
"My name is {name} and age is {age}".format(name="Joker",age=25)
```

*  we can call attributes , methods inside {}
*  we can use square brackets to use offsets for lists or tuples , keys for dictionary
*  we can use * unpack values in tuple
*  Sadly negative indexes and slicing does not work in format method

```python
"My details is {0[name]} and I m {0[age]}".format({"name":"Geaorge","age":25})
```

```python
"{0[0]}".format([100,200,300])
```

```python
"{name[1]}".format(name=[1,2,34])
```

```python
"{name[-1]}".format(name=[1,2,3]) # so formatting does not supports index be negative
```

```python
"{name[:]}".format(name=[1,2,3]) # format method does not support slicing
```

```python
num = 1,2,3,4,5
"{},{},{},{},{}".format(*num) # unpack tuples
```

[keyname or numbers][attributes or indexes][!type changing thing]:[formatSpec]
* keyname or numbers is that 0 or "name"
* indexes or attributes
* conversion flag: !s , !r how it represents
* formatspec : specifies the format

[fill][align][sign][width][seperator].[precision]

* fill : fills with something
* align : aligns left or right
* sign : sign of digit
* width : width of number
* precision how many numbers after decimal points

```python
# fill with zeros
"{0:012}".format(1234.56) # increases the width
```

```python
"{0:0>12}".format(1234.56) # right align
```

```python
"{0:0<12}".format(1234.56) # left align
```

```python
"{0:0^12}".format(1234.56) # center align
```

```python
"{0:,.2f}".format(12_0000.32)
```

```python
"{0:+,.5f}".format(-12_00_000.32456)
```

```python
"{0:{1}.{2}f}".format(123.56,'+',5)
```
