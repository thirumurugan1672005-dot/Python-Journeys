# How to Define Dictionary
## Create Traditional form

```python
person = {'name':'user','age':24}
print(person)
```

## Create Dictionary Dynamically

```python
person = {}
person['name'] = 'user'
person['age'] = 24
print(person)
```

## Create Dictionary using key value arguments (keys are Strings)

```python
person = dict(name='user',age=24)
print(person)
```

## Create Dictionary using list of key-value pairs

```python
person = dict([('name','user'),('age',24)])
print(person)
```

## Create Dictionary from pair of lists

```python
person = dict(zip(['name','age'],['user',25]))
print(person)
```

## Create using fromkeys method

```python
person = dict.fromkeys(['name','age'],'one')
print(person)
```

# Dictionary Unpacking 

```python
D = dict(a=4, c=3,**dict(d=4))
D
```

```python
# dict does not accepts duplicate keys
D = dict(a=4, c=3,**dict(d=4,c=25))
D
```

# Dictionary Comprehensions
* Dictionary Comprehensions are the comprehensions like lists
* It is way to express dictionary in comprehensions
* use zip to make the dictionary 

```python
a = zip(['a','b','c'],[1,2,3])
b = dict(a)
b
```

```python
# comphrension
fruits = ['apple','oranges','guava']
prices = [260 , 180 , 45]
market = {fruit:price for fruit , price in zip(fruits,prices)}
print(market)
```
