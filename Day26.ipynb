# Shared References with Mutable objects
* Mutable objects are the objects which supports inplace changes
* When both variables shared the same object 
* Inplace change in one affects another 
* avoid by take slice or copy() method
* nested ones use deepcopy() method

````python
L1 = [1,2,3]
L2 = L1 
L1[0] = 24
print(L2)
````

````python
L1 = [1,2,3]
L2 = L1.copy()
L3 = L1[:]
L1[0] = 24
print(L2,L3)
````

````python
L1 = [[1,2,3],[4,5,6]]
L2 = L1.copy()
L3 = L1[:]
L1[0][1] = 24
print(L2,L3,sep="\n")
````

````python
from copy import deepcopy
L1 = [[1,2,3],[4,5,6]]
L2 = deepcopy(L1)
L1[0][1] = 24
print(L2)
````
