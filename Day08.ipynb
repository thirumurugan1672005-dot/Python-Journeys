# Decimal : Fixed Precision objects
* Decimal fixed Precision objects are just like floating Point numbers but with fixed precision
* It can be constructed from string or float
* It has peformance penalty for this

```python
import decimal
decimal.Decimal('0.1')
```

```python
import decimal
exp = decimal.Decimal('0.1') + decimal.Decimal('0.2')
print(exp)
```

Rememeber passing float means you are passing same underlying precision issues

```python
import decimal
exp = decimal.Decimal(0.1) + decimal.Decimal(0.2)
print(exp) 
```

```python
import decimal
exp = decimal.Decimal('0.2345')
print(exp)
decimal.getcontext().prec = 2
exp = decimal.Decimal('0.2') + decimal.Decimal("0.34565")
print(exp)
```

```python
A = {1,2,3,4,5,6}
B = {2,3,5,7,8,9}
print(A.union(B) , A | B)
print(A.intersection(B),A & B)
print(A.difference(B) , A - B)
print(A.symmetric_difference(B), A ^ B)
print(A.issubset(B) , A < B)
```

```python
A = {1,2,3}
B = {1,2,3,4,5}
print(A < B , A.issubset(B))
```

```python
A , B = {1,2,3},{1,2,3,4,5}
B > A
```

```python
# update or |= 
A = {1,2,3}
B = {1,2,3,4,5}
C = B
B |= A
print(B)
C.update(A)
print(C)
```

```python
# intersection_update or &=
A = {1,2,3,4,6}
B = {1,2,3,4,5}
C = B
B &= A
print(B)
C.intersection_update(A)
print(C)
```

```python
# symmetric_diiference_update or ^=
A = {1,2,3,4,6}
B = {1,2,3,4,5}
C = B
B = B ^ A
print(B)
C.symmetric_difference_update(A)
print(C)
```

```python
# diiference_update or &=
A = {1,2,3,4,6}
B = {1,2,3,4,5}
C = B
B = B - A
print(B)
C.difference_update(A)
print(C)
```

```python
# issubset or <
A = {1,2,3,4}
B = {1,2,3,4,5,6}
print(A < B)
print(A.issubset(B))
```

```python
# issuperset or >
A = {1,2,3,4}
B = {1,2,3,4,5,6}
print(A > B)
print(A.issuperset(B))
```
