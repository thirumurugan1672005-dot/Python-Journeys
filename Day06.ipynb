# Comparison Operators
* Comparison Operators are used to compare the values
* Complex numbers cannot be compared with integers and floating
* floating and integers are compared

````python
a = 12 
b = 12.3
a < b
````

````python
exp1 , exp2 = 12.34 , 34.56
print(exp1 >= exp2)
````

# Comparison Operator types
1. ```>``` Greater than
2. < Lesser than
3. ```>=``` Greater or equal to
4. <= Lesser or equal to
5. == Equality
6. != Not Equality

# Chained Comparisons
Chained Comparisons chained based with results A < B < C

````python
A = 12
B = 24
C = 36
A < B < C
````

````python
C < A > B # (C < A) and (A > B)
````

# Floor vs Truncation
Floor means smallest integer closest to true result

````python
# Floor
import math
a = 12 // 3
print(a)
b = -12 // 3.5
print(b)
print(math.floor(12//3))
print(math.floor(-12//3.5))
````

````python
# Truncate
import math
exp = math.trunc(12//3.5)
print(exp)
print(12 / 3.8)
````

# Hexal , Octal , Binary
## Hexadecimal
Hexadecimal is the representation of integer in hex form where 0-9 A-F is used
## Binary
Where 0 and 1 are used
## Octal
where 0-7 used

* For hexadecimal starts with 0x or 0X
* For octal starts with 0o or 0O
* for binary starts with 0b or 0B

````python
exp = 0xFFF
print(exp)
exp = 0xF
print(exp)
````

````python
exp = 0O777777
print(exp)
````

````python
exp = 0b1000
print(exp)
````
