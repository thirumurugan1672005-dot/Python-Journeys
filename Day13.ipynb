# F-Strings
* F-Strings are the strings in which words inside curly braces substituted from variables
* It is runtime expression in which it is substituted

```python
# formatting just like format method 

salary = 1234.56
f"{salary:,.2f}"
```

```python
salary = 123.5678
f"{salary:.3f}"
```

```python
num = 12 # binary
f"{num:b}"
```

```python
num = 2087 # hex
f"{num:x}"
```

```python
num = 3012 # octal
f"{num:o}"
```

 F Strings also create the new string

```python
name = 'hello'
f"{name!r}"
```

```python
name = 'hello'
f"{name!a}"
```

```python
name = 'hello'
f"{name!s}"
```
