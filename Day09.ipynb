 # Introduction to Strings 
 * Strings are textual information or binary files
 * Strings are sequences they are ordered from left to right and accessed by positions
 * They are sequence of one character strings

## Sqeuence Operations
* The Operations peformed generally by sequences
* Concatenation and Repetiton

```python
exp = "Hello"
print(exp)
# Concatenation
print("Hello"+"World")
# Repetition
exp = "Py" * 3
print(exp)
```

## Indexing
* Indexing refers to the [] to access elements in sequence
* It bounds to length of sequence - 1
* Able to access elements
* From left to right (Positive indexes)
* Negative indexes are used to traverse from right to left

```python
exp = "Python"
print(exp[0],exp[-len(exp)])
print(exp[-1],exp[len(exp)-1])
```

```python
# OUT OF BOUNDS RESULTS IN ERROR
exp = "Python"
exp[23]
```

# Slicing 
* Slicing is the form [X:Y] while starts from X offset to Y offset - 1
* It bounds to length of string

```python
exp = "Papaya"
exp[:]
```

```python
# X : normally 0 
exp[:3]
```

```python
# Y normally length of list 
exp[2:]
```

```python
exp[2:4]
```

* Above said are true when third is postive default
* [X:Y:Z] here Z = 1(step) default
* positive step : left -> right
* negative step : right -> left
* Default X: 0 , Y: len(sequence) when step is positive
* Default X:-1 , Y:-len(sequence)-1 when step is negative

```python
exp = "PapayaTree"
exp[::-1]
```

```python
exp[-1:-len(exp)-1:-1]
```

## Immutability
* Immutability refers to the content inside cannot be changed
* We can assign only new objects
* Strings are immutable
* Python is efficient enough to use garbage collector to clean memory spaces if the object is left out

```python
exp = "Python"
exp[0] = "P"
```

```python
exp = "Python"
print(exp)
exp = "Java"
print(exp)
```

## Iterating through Strings using for loop

```python
statement = "Today is the best day i ever have"
for st in statement:
    print(st,end = " ")
```
